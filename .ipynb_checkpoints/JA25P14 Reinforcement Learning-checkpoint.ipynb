# Reinforcement Learning
Reinforcement learning (RL) is one of three machine learning paradigms, alongside supervised learning and unsupervised learning. RL differs from supervised learning as it focus on finding a balance between exploration and exploitation.

In this approach, the (algorithm) agents take actions in an environment in order to maximize its cumulative reward. The environment is typically stated in the form of a Markov decision process (MDP), because many reinforcement learning algorithms for this context utilize dynamic programming techniques.

MDP is a discrete-time stochastic control process and provides a mathematical framework for modeling decision making in situations where outcomes are partly random and partly under the control of a decision maker. Such processes are useful in optimization problems solved via dynamic programming and reinforcement learning.

RL framework is extremely promising in trading and investment management. Currently RL is mostly a research area and hasn’t yet had significant practical successes beyond games. However, it’s an idea whose time has come. 

**Fundamental Notions**

Following are some of the fundamental notions of the RL

<img src="https://www.kdnuggets.com/images/reinforcement-learning-fig1-700.jpg" width="40%" align="center" style="vertical-align: bottom;">

**Environment :** The world through which agents moves. The environment defines the problem at hand. For example, the environment can be a financial market to be traded in. 

**State :** Current condition returned by the environment. A state subsumes all relevant parameters that describe the current state of the environment. In a financial market, this might include current and historical price levels or financial indicators such as moving averages, macroeconomic variables, and so on.

**Agent :** The RL algorithm that learns from the exploration and exploitation. Agent subsumes all elements of the RL algorithm that interacts with the environment and learns from these interactions. In a financial context, the agent could represent a trader placing bets on rising or falling markets.

**Action :** All possible steps that the agent can take from a (limited) set of allowed actions. In a financial market, going long or short could be admissible actions.

**Reward :** An instant return from the environment to appraise the last action. In a financial context, profit (or loss) is a standard reward.

**Policy :** The approach the agent uses to determine the next action based on the current state. In a financial context, a trading bot that observes three price increases in a row might decide to buy the market.

**Value :** The expected long-term return with discount, as opposed to short-term rewards. For a financial trading bot, this might be the expected accumulated trading profit.

**Episode :** The set of steps from the initial state of the environment until success is achieved or failure is observed. In the financial world, this might be from the beginning of the year to the end of the year or to bankruptcy.

# OpenAI Gym

OpenAI [<sup>[2]</sup>](#fn2) is an organization that facilitate research in RL. OpenAI has developed and open sourced a suite of environments, called **`OpenAI Gym`**, that allows the training of RL agents via a standardized API.

The gym library addressed two important gaps that were slowing down the RL research.

1. Need for better benchmarks
2. Lack of standardization of environments used in publications

Gym is a toolkit for developing and comparing RL algorithms and makes no assumptions about the structure of the agent. It is compatible with any numerical computation library, such as TensorFlow or Theano. The gym library is a collection of environments that we can use to work out the RL algorithms. These environments have a shared interface, allowing us to write general algorithms.

## Stable-Baselines3

Stable-Baselines3 (SB3) [<sup>[3]</sup>](#fn3) is a popular, open-source library that provides implementations of RL algorithms in Python. It is a successor of the original Stable-Baselines library and is designed to be user-friendly, modular, and well-documented for both research and practical applications in reinforcement learning.

SB3 is built on top of PyTorch, making it a flexible and powerful tool for developing and testing RL algorithms. The library provides standardized implementations of state-of-the-art RL algorithms, which allows users to apply these techniques to different types of environments without having to implement the algorithms from scratch.

SB3 is to reinforcement learning what scikit learn is to machine learning. It provides a simple and intuitive interface for training and evaluating RL agents, making it easy to experiment with different algorithms and configurations.

### Defining Environment

The StockTradingEnv class is defined to represent the stock trading environment. It inherits from the gym.Env class and implements the __init__, reset, and step methods. The __init__ method initializes the environment with the stock data, reward range, action space, observation space, and other parameters. The reset method resets the environment to its initial state, and the step method executes the action and returns the observation, reward, done flag, truncated flag, and info.

```python
# Stock Trading Environment
class StockTradingEnv(gym.Env):
    # your code here
    self.reward_range = (-np.inf, np.inf)
    self.action_space = spaces.Discrete(3)  # Buy, Sell, Hold
    self.observation_space = spaces.Box(low=0, high=1, shape=(6,), dtype=np.float32)
```

The reward range is defined in the `reward_range` attribute of our StockTradingEnv class. In this case, the reward range is set to (-np.inf, np.inf), which means that the reward can be any value. This is because we are using a continuous reward space, and we want to allow the agent to take any action and receive any reward.

The action space is defined in the `action_space` attribute of our StockTradingEnv class. In this case, we have three actions: Buy, Sell, and Hold. The spaces.Discrete class is used to define the action space as a discrete set of actions. This means that the agent can only take one of the three actions at a time.

The observation space is defined in the `observation_space` attribute of our StockTradingEnv class. In this case, we have a vector of 6 elements, where each element represents a different aspect of the stock or position. For example, the first element of the vector will represent the current balance of the agent, the second element will represent the number of shares held, and so on. The parameter in `spaces.Box` defines the structure of the observation space in our reinforcement learning environment. `low=0` and `high=1` specifies that all values in the observation vector will be between 0 and 1. This is why we normalize all our values in the `_next_observation` method.

### Step Function

In gymnasium (newer version of OpenAI Gym), step() function now returns five values instead of the four used in the older gym library. Observing the outcome of our action is critical for us to do better than random actions. The environment’s step function returns exactly what we need by returning the following five values.

``` python
# Stock Trading Environment
class StockTradingEnv(gym.Env):
    # your code here
    def step(self, action):
        # your code here
```

`observation`: Typically a NumPy array or another data structure that describes the current state of the environment. It represents what the agent sees at the current time step, based on the environment.

`reward`: A floating-point number that specifies the amount of reward returned as a result of taking the action. The scale varies between environments, but the goal is always to increase your total reward.

`terminated: A Boolean (True or False). The terminated flag indicates whether the agent has reached a terminal state, meaning the episode is over because a goal condition or natural ending has been reached. When it is True, the agent knows it has reached the terminal state, and the episode is finished.

`truncated`: A Boolean (True or False). The truncated flag indicates whether the episode was ended prematurely, typically due to a time limit or some kind of constraint rather than achieving a natural terminal state. This can happen if the agent takes too long to learn the environment’s dynamics, or if the environment is too difficult to solve. In many cases, you may not have a truncation condition and this flag can be ignored or set to False.

`info`: A dictionary that contains auxiliary diagnostic information helpful for debugging, and sometimes learning. For example, it might contain the raw probabilities behind the environment’s last state change. However, official evaluations of your agent are not allowed to use this for learning.

The process gets started by calling `reset()`, which returns an initial `observation` or state. So a more proper way of writing the code would be to respect the `terminated` flag as the observation (state) of the environment is given by these five parameters.



## Full Code

````python
# Stock Trading Environment
class StockTradingEnv(gym.Env):
    def __init__(self, df,  initial_balance=100000, max_shares=100):
        super().__init__()  # Correct initialization
        self.df = df
        self.reward_range = (-np.inf, np.inf)
        self.action_space = spaces.Discrete(3)  # Buy, Sell, Hold
        self.observation_space = spaces.Box(low=0, high=1, shape=(6,), dtype=np.float32)

        self.current_step = 0
        self.total_steps = len(df)
        self.initial_balance = initial_balance
        self.balance = self.initial_balance
        self.shares_held = 0
        self.current_price = 0
        self.max_shares = max_shares
        self.cumulative_rewards = []  # Track cumulative rewards

    # Reset the environment
    def reset(self, seed=2024):
        super().reset(seed=seed)  # Correct reset method with seed
        self.current_step = 0
        self.balance = self.initial_balance
        self.shares_held = 0
        self.current_price = self.df.iloc[self.current_step]['Close']
        self.cumulative_rewards = []  # Reset rewards on new episode
        return self._next_observation(), {}

    # Return the next observation
    def _next_observation(self):
        frame = np.array([
            self.balance / self.initial_balance,
            self.shares_held / self.max_shares,
            self.current_price / self.df['Close'].max(),
            self.df.iloc[self.current_step]['Open'] / self.df['Open'].max(),
            self.df.iloc[self.current_step]['High'] / self.df['High'].max(),
            self.df.iloc[self.current_step]['Low'] / self.df['Low'].max(),
        ])
        return frame

    # Step the environment
    def step(self, action):
        # Store the previous portfolio value for reward calculation
        prev_portfolio_value = self.balance + self.shares_held * self.current_price

        self.current_step += 1
        self.current_price = self.df.iloc[self.current_step]['Close']

        # Execute the action
        if action == 0:  # Buy
            shares_to_buy = min(self.balance // self.current_price, self.max_shares - self.shares_held)
            self.balance -= shares_to_buy * self.current_price
            self.shares_held += shares_to_buy
        elif action == 1:  # Sell
            shares_to_sell = self.shares_held
            self.balance += shares_to_sell * self.current_price
            self.shares_held = 0

        # Hold does nothing

        # Calculate the new portfolio value after the action
        new_portfolio_value = self.balance + self.shares_held * self.current_price

        # Reward is the change in portfolio value
        reward = new_portfolio_value - prev_portfolio_value

        # Track cumulative rewards (this is for later plotting)
        self.cumulative_rewards.append(reward)

        # Check if episode is done
        terminated = self.current_step >= self.total_steps - 1

        # For simplicity, we're not truncating the episode in this environment
        truncated = False

        # Return observation, reward, done flag, and info
        obs = self._next_observation()
        return obs, reward, terminated, truncated, {}


# Train the agent 
def train_agent(env, model_class, model_params, train_timesteps):
    model = model_class(policy="MlpPolicy", env=env, **model_params, tensorboard_log=logdir)
    model.learn(total_timesteps=train_timesteps, tb_log_name='PPO')
    
    # To save the trained model:
    # model.save("ppo_trading_model")
    
    # To load the trained model later:
    # model = PPO.load("ppo_stock_trading_model")
    
    return model


# Test the trained agent on test data
def test_agent(env, model):
    obs, _ = env.reset()    
    for _ in range(len(env.df) - 1):
        action, _ = model.predict(obs)
        obs, reward, terminated, truncated, _ = env.step(action)

        if terminated:
            break

    return np.cumsum(env.cumulative_rewards)

# Plot cumulative rewards
def plot_cumulative_rewards(rewards, name):
    plt.plot(rewards, label=name)
    plt.xlabel('Trading Days')
    plt.ylabel('Cumulative Reward')
    plt.title(f'Cumulative Reward - {name}')
    plt.legend()
    plt.show()
````

### Train & Test
Now that we have our environment and model ready, we can train the model and evaluate its performance on the test dataset. Download the dataset from [here](https://raw.githubusercontent.com/kannansingaravelu/datasets/refs/heads/main/niftyindex.csv) and place it in the data folder for access.

````python
# Main function
def main():
    # Get stock data
    df = pd.read_csv('./data/niftyindex.csv', index_col=0, parse_dates=True)
    print("Data loaded successfully")

    # Split the data into training and test sets
    ratio = 0.80
    train_size = int(df.shape[0]*ratio)
    train, test = df.iloc[:train_size], df.iloc[train_size:] 
    print("Data split successfully")
    print(f"Traning Data: {train.shape}, Test Data: {test.shape}")

    # Initialize the training & test environment
    train_env = StockTradingEnv(train)
    test_env = StockTradingEnv(test)
    print("Environments initialized successfully")

    # Define RL algorithm
    algorithm = {
        "PPO": (A2C, {"learning_rate": 0.0003}),
        # "A2C": (A2C, {"learning_rate": 0.0003}),
    }
    
    # Train, test, and plot for each algorithm
    for name, (algo, params) in algorithm.items():
        print("Training the agent...")
        model = train_agent(train_env, algo, params, train_timesteps=100000)
        # rewards_train = np.cumsum(train_env.cumulative_rewards)

        # print("Testing the agent...")
        rewards = test_agent(test_env, model)
        

        # Plot the cumulative rewards for each algorithm
        plot_cumulative_rewards(rewards, name)
    
if __name__ == "__main__":
    main()
````
